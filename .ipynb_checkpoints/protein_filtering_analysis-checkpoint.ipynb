# Protein Database Filtering Analysis

This notebook filters protein data through multiple stages:
1. Filter for all-alpha proteins
2. Filter for proteins with less than n residues
3. Filter for proteins with lowest sequence similarity

Each stage reports the number of proteins before and after filtering.

```python
import pandas as pd
import sqlite3
from sqlalchemy import create_engine, Column, Integer, String, Float, text
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.orm import sessionmaker
import os
from pathlib import Path

Base = declarative_base()

class Protein(Base):
    __tablename__ = 'proteins'
    
    id = Column(Integer, primary_key=True)
    pdb_chain = Column(String, nullable=False)
    protein_type = Column(String, nullable=False)
    length = Column(Integer, nullable=False)
    similarity = Column(Float, nullable=False)
    source = Column(String)

print("Imports and model setup complete")
```

```python
# Create database engine
engine = create_engine('sqlite:///protein_analysis.db', echo=False)
Base.metadata.create_all(engine)

Session = sessionmaker(bind=engine)
session = Session()

print("Database setup complete")
```

```python
# Load protein data from metadata files
metadata_files = [
    'protein_selection/datasets/alpha/metadata.tsv',
    'protein_selection/datasets/beta/metadata.tsv'
]

all_proteins = []

for metadata_file in metadata_files:
    if os.path.exists(metadata_file):
        print(f"Loading data from {metadata_file}")
        df = pd.read_csv(metadata_file, sep='\t')
        print(f"Loaded {len(df)} proteins from {metadata_file}")
        
        for _, row in df.iterrows():
            protein = Protein(
                pdb_chain=row['PDB_Chain'],
                protein_type=row['Type'],
                length=row['Length'],
                similarity=row['Similarity'],
                source=row.get('Source', 'unknown')
            )
            all_proteins.append(protein)
    else:
        print(f"Warning: {metadata_file} not found")

# Add all proteins to database
session.add_all(all_proteins)
session.commit()

total_proteins = session.query(Protein).count()
print(f"\nTotal proteins loaded into database: {total_proteins}")
```

```python
# Get initial protein counts by type
print("=== Initial Protein Database Statistics ===")

total_count = session.query(Protein).count()
print(f"Total proteins in database: {total_count}")

# Count by protein type
type_counts = session.query(Protein.protein_type, text('COUNT(*)')).group_by(Protein.protein_type).all()
for protein_type, count in type_counts:
    print(f"{protein_type} proteins: {count}")

print("\n" + "="*50)
```

```python
# Stage 1: Filter for all-alpha proteins
print("=== Stage 1: Filtering for All-Alpha Proteins ===")

before_alpha_count = session.query(Protein).count()
print(f"Proteins before alpha filter: {before_alpha_count}")

# Apply alpha filter
alpha_proteins = session.query(Protein).filter(Protein.protein_type == 'alpha').all()
after_alpha_count = len(alpha_proteins)

print(f"Proteins after alpha filter: {after_alpha_count}")
print(f"Proteins filtered out: {before_alpha_count - after_alpha_count}")
print(f"Percentage remaining: {(after_alpha_count/before_alpha_count)*100:.2f}%")

# Store alpha proteins for next stage
current_proteins = alpha_proteins
print("\n" + "="*50)
```

```python
# Stage 2: Filter for proteins with less than n residues
# You can adjust this value as needed
max_residues = 100  # Change this value as desired

print(f"=== Stage 2: Filtering for Proteins with < {max_residues} Residues ===")

before_residue_count = len(current_proteins)
print(f"Proteins before residue filter: {before_residue_count}")

# Apply residue length filter
length_filtered_proteins = [p for p in current_proteins if p.length < max_residues]
after_residue_count = len(length_filtered_proteins)

print(f"Proteins after residue filter: {after_residue_count}")
print(f"Proteins filtered out: {before_residue_count - after_residue_count}")
print(f"Percentage remaining: {(after_residue_count/before_residue_count)*100:.2f}%")

# Store filtered proteins for next stage
current_proteins = length_filtered_proteins
print("\n" + "="*50)
```

```python
# Stage 3: Filter for proteins with lowest sequence similarity
# You can adjust this threshold as needed
similarity_threshold = 0.3  # Change this value as desired

print(f"=== Stage 3: Filtering for Proteins with Similarity < {similarity_threshold} ===")

before_similarity_count = len(current_proteins)
print(f"Proteins before similarity filter: {before_similarity_count}")

# Apply similarity filter
similarity_filtered_proteins = [p for p in current_proteins if p.similarity < similarity_threshold]
after_similarity_count = len(similarity_filtered_proteins)

print(f"Proteins after similarity filter: {after_similarity_count}")
print(f"Proteins filtered out: {before_similarity_count - after_similarity_count}")
print(f"Percentage remaining: {(after_similarity_count/before_similarity_count)*100:.2f}%")

# Final filtered proteins
final_proteins = similarity_filtered_proteins
print("\n" + "="*50)
```

```python
# Final Summary
print("=== FILTERING SUMMARY ===")
print(f"Initial proteins: {total_count}")
print(f"After alpha filter: {after_alpha_count} ({(after_alpha_count/total_count)*100:.2f}%)")
print(f"After residue filter (< {max_residues}): {after_residue_count} ({(after_residue_count/total_count)*100:.2f}%)")
print(f"After similarity filter (< {similarity_threshold}): {after_similarity_count} ({(after_similarity_count/total_count)*100:.2f}%)")
print(f"\nTotal proteins removed: {total_count - after_similarity_count}")
print(f"Final retention rate: {(after_similarity_count/total_count)*100:.2f}%")
```

```python
# Additional Analysis: Show distribution of final proteins
if final_proteins:
    print("=== Final Protein Set Analysis ===")
    
    lengths = [p.length for p in final_proteins]
    similarities = [p.similarity for p in final_proteins]
    
    print(f"Length statistics:")
    print(f"  Min: {min(lengths)} residues")
    print(f"  Max: {max(lengths)} residues")
    print(f"  Average: {sum(lengths)/len(lengths):.2f} residues")
    
    print(f"\nSimilarity statistics:")
    print(f"  Min: {min(similarities):.4f}")
    print(f"  Max: {max(similarities):.4f}")
    print(f"  Average: {sum(similarities)/len(similarities):.4f}")
    
    print(f"\nFirst 10 final proteins:")
    for i, protein in enumerate(final_proteins[:10]):
        print(f"{i+1:2d}. {protein.pdb_chain:15s} | Length: {protein.length:3d} | Similarity: {protein.similarity:.4f}")
else:
    print("No proteins passed all filters with current thresholds.")
```

```python
# Demonstrate SQLAlchemy queries for each filter stage
print("=== SQLAlchemy Query Examples ===")

# Stage 1: All-alpha proteins using SQLAlchemy
alpha_query = session.query(Protein).filter(Protein.protein_type == 'alpha')
alpha_count_sqlalchemy = alpha_query.count()
print(f"Alpha proteins (SQLAlchemy query): {alpha_count_sqlalchemy}")

# Stage 2: Alpha proteins with < max_residues using SQLAlchemy
residue_query = alpha_query.filter(Protein.length < max_residues)
residue_count_sqlalchemy = residue_query.count()
print(f"Alpha + residue filter (SQLAlchemy query): {residue_count_sqlalchemy}")

# Stage 3: All filters combined using SQLAlchemy
final_query = residue_query.filter(Protein.similarity < similarity_threshold)
final_count_sqlalchemy = final_query.count()
print(f"All filters combined (SQLAlchemy query): {final_count_sqlalchemy}")

# Show the actual SQL that would be generated
print(f"\nGenerated SQL for final query:")
print(str(final_query.statement.compile(compile_kwargs={"literal_binds": True})))
```

```python
# Cleanup
session.close()
print("Session closed.")
```
